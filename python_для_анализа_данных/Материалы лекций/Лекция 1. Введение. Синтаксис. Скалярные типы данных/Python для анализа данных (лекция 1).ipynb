# Переменные и передача аргументов

Присваивание значения переменной (или имени) в Python приводит к созданию ссылки на объект, стоящий в правой части присваивания. Рассмотрим список целых чисел:

```python
a = [1, 2, 3, [1, 2]]
# Предположим, что мы присвоили значение a новой переменной b:
b = a.copy()
a[-1].append(4)
print(a)
print(b)
```

```python
b.append(4)
a
```

Понимать семантику ссылок в Python и знать, когда, как и почему данные копируются, особенно важно при работе с большими наборами данных.

# Импорт

Импорт в Python позволяет использовать функции, классы и модули, определенные в других файлах или библиотеках. Это важный инструмент для организации кода и повторного использования кода. В Python модуль – это просто файл с расширением .py, который содержит функции и различные определения, в том числе импортированные из других py-файлов. Пусть имеется следующий модуль:

```python
# some_module.py
%%writefile some_module.py

PI = 3.14159

def f(x):
    return x + 2

def g(a, b):
    return a + b
```

Если бы мы захотели обратиться к переменным или функциям, определенным в some_module.py, из другого файла в том же каталоге, то должны были бы написать:

```python
import some_module

result = some_module.f(5)
pi = some_module.PI
print(result, pi, sep = '\n')
```

Ключевое слово as позволяет переименовать импортированные сущности:

```python
import some_module as sm
from some_module import PI as pi, g as gf

r1 = sm.f(pi)
r2 = gf(6, pi)

print(r1, r2, sep = '\n')
```

# Проверка ссылок на объект



Для проверки того, что две ссылки ведут на один и тот же объект, служит оператор is. Оператор is not тоже существует и позволяет проверить, что два объекта различаются.

```python
a = [1, 2, 3]
b = a
c = list(a)
print(a is b)
print(a is not c)
```

Поскольку функция list всегда создает новый список Python (т. е. копию исходного), то есть уверенность, что c и a – различные объекты. Сравнение с помощью оператора is – не то же самое, что с помощью оператора ==, потому что
в данном случае мы получим:

```python
a = [1, 2, 3]
c = list(a)
a == c
```

Операторы is и is not часто употребляются, чтобы проверить, равна ли некоторая переменная None, потому что существует ровно один экземпляр None:

```python
a = None
a is None
```

# if, elif, else

Предложение if – одно из самых хорошо известных предложений управления потоком выполнения. Оно вычисляет условие и, если получилось True, исполняет код в следующем далее блоке:

```python
x = -5
if x < 0:
    print("Отрицательно")
```

После предложения if может находиться один или несколько блоков elif и блок else, который выполняется, если все остальные условия оказались равны False:

```python
x = 3
if x < 0:
    print("Отрицательно")
elif x == 0:
    print("Равно нулю")
elif 0 < x < 5:
    print("Положительно, но меньше 5")
else:
    print("Положительно и больше или равно 5")
```

Если какое-то условие равно True, последующие блоки elif и else даже не рассматриваются. В случае составного условия, в котором есть операторы and или or, частичные условия вычисляются слева направо с закорачиванием:

```python
a = 5; b = 7
c = 8; d = 4
if a < b or c > d:
    print("Сделано")
```

В этом примере условие c > d не вычисляется, потому что уже первое сравнение a < b равно True.

Можно также строить цепочки сравнений:

```python
4 > 3 > 2 > 1
```

# Логические операторы: and, or, not

В Python логические операторы работают по принципу "ленивого вычисления" (short-circuit evaluation)

**Оператор and**

Поведение:

*   Возвращает первый ложный операнд
*   Если все операнды истинны, возвращает последний операнд


```python
# Базовые случаи
print(True and True)    # True
print(True and False)   # False
print(False and True)   # False
print(False and False)  # False

# С небулевыми значениями
print(5 and 10)         # 10 (оба истинны, возвращает последний)
print(0 and 10)         # 0 (первый ложный)
print("hello" and "")   # "" (первый истинный, второй ложный)
print(None and "world") # None (первый ложный)
```

**Оператор or**

Поведение:

*   Возвращает первый истинный операнд
*   Если все операнды ложны, возвращает последний операнд






```python
print(True or True)     # True
print(True or False)    # True
print(False or True)    # True
print(False or False)   # False

# С небулевыми значениями
print(5 or 10)          # 5 (первый истинный)
print(0 or 10)          # 10 (первый ложный, второй истинный)
print("" or "hello")    # "hello" (первый ложный, второй истинный)
print(None or 0 or 42)  # 42 (первые два ложные)
```

```python
# Установка значения по умолчанию
config = None
port = config or 8080
print(port)  # 8080

username = ""
display_name = username or "Гость"
print(display_name)  # "Гость"

# Проверка нескольких условий
age = 17
if age < 18 or age > 65:
    print("Специальная категория")
```

**Оператор not**

Поведение:



*   Всегда возвращает булевое значение (True или False)

*   Инвертирует логическое значение операнда





```python
print(not True)   # False
print(not False)  # True

# С небулевыми значениями
print(not 5)      # False (5 - истина)
print(not 0)      # True (0 - ложь)
print(not "hello") # False (непустая строка - истина)
print(not "")     # True (пустая строка - ложь)
```

```python
items = []
if not items:
    print("Список пуст")  # Выполнится

user_logged_in = False
if not user_logged_in:
    print("Пожалуйста, войдите в систему")  # Выполнится

# Комбинация с другими операторами
x = 10
if not (x < 0 or x > 100):
    print("x в допустимом диапазоне")  # Выполнится
```

**Порядок выполнения** (от высшего к низшему):



1.   not
2.   and
3.   or

```python
# Пример с приоритетами
result = not False and True or False
# Вычисляется как: (not False) and True or False
# → True and True or False
# → True or False
# → True
print(result)  # True

# Изменение приоритета скобками
result2 = not (False and (True or False))
# → not (False and True)
# → not False
# → True
print(result2)  # True
```

# Циклы for и while

```python
Циклы for предназначены для обхода коллекции (например, списка или кортежа) или итератора. Стандартный синтаксис выглядит так:
```

```python
for value in collection:
    # что-то сделать с value
```

Ключевое слово continue позволяет сразу перейти к следующей итерации цикла, не доходя до конца блока. Рассмотрим следующий код, который суммирует целые числа из списка, пропуская значения None:

```python
sequence = [1, 2, None, 4, None, 5]
total = 0
for value in sequence:
    if value is None:
        continue
    total += value
total
```

Ключевое слово break осуществляет выход из самого внутреннего цикла, объемлющие циклы продолжают работать:

```python
for i in range(4):
    for j in range(4):
        if j > i:
            break
        print((i, j))
```

Цикл while состоит из условия и блока кода, который выполняется до тех пор, пока условие не окажется равным False или не произойдет выход из цикла в результате предложения break:

```python
x = 256
total = 0
iterat = 0
while x > 0:
    if total > 500:
        break
    total += x
    x = x // 2
    iterat += 1
print(total, iterat)
```

# Функция range

Функция range порождает последовательность равноотстоящих целых чисел:

```python
range(10, 90)
```

```python
list(range(10, 20, 2))
```

Можно задать начало и конец диапазона и шаг (который может быть отрицательным):

```python
list(range(0, 20, 2))
```

```python
list(range(5, 0, -1))
```

Как видите, конечная точка в порождаемый range диапазон не включается.
Типичное применение range – обход последовательности по индексу:

```python
seq = [1, 2, 3, 4]
for i in range(len(seq)):
    print(f"элемент с индексом {i}: {seq[i]}")
```

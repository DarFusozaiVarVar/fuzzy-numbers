# Создание новых столбцов на основе существующих

```python
#Записываем расчеты по формуле в столбец df_od['Discount_fact']. Этот столбец создается в момент выполнения кода
#формула расчитывает скидку фактическую, Discount - это скидка в процентах, точнее в долях
df_od['Discount_fact']= df_od['UnitPrice']*df_od['Quantity']*df_od['Discount']
```

```python
#расчитываем оборот очищенный от фактической скидки
df_od['Revenue']= df_od['UnitPrice']*df_od['Quantity']-df_od['Discount_fact']
```

```python
#выводим данные с датафрема. Появились 2 новых столбца
df_od.head()
```

# Типы данных

```python
#возможные типы данных:

#object - текст
#int64 - целые числа
#float64 - числа с плавающей точкой
#datetime64 - дата / время
#bool - содержит данные типа True / False
```

```python
#выводим данные с датафрейма
df_o
```

```python
#каждый столбец в пределах датафрейма имеет собственный тип данных
df_o.dtypes
```

```python
#даты в датафрейме df_o изначально имеют тип данных object, но правильнее поменять их тип данных на datetime64
#для этого есть метод pd.to_datetime()
df_o['OrderDate'] = pd.to_datetime(df_o['OrderDate'])
df_o['RequiredDate'] = pd.to_datetime(df_o['RequiredDate'])
df_o['ShippedDate'] = pd.to_datetime(df_o['ShippedDate'])
```

```python
#убеждаемся, что у дат поменялся тип данных
df_o.dtypes
```

```python
#менять тип данных можно также методом astype()
#пример перевода типа данных int64 в object
df_o['OrderID']= df_o['OrderID'].astype('object')
df_o.dtypes
```

```python
# Параметр parse_dates позволяет автоматически преобразовать указанные столбцы в datetime непосредственно при чтении файла.
# Укажите список названий столбцов, которые нужно преобразовать
```

```python
df_o = pd.read_csv('/content/drive/MyDrive/Orders.csv', sep = ';', parse_dates=['ShippedDate', 'OrderDate'])
df_o.dtypes
```

```python
#даты с типом данных datetime64 можно отнимать
df_o['delivery_time']= df_o['ShippedDate']- df_o['OrderDate']
```

```python
#появился новый столбец delivery_time, в котором записана разница в днях
df_o.head()
```

```python
#у столбца delivery_time тип данных timedelta64[ns]
df_o.dtypes
```

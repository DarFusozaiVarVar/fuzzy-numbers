# Фильтрация строк по одному условию


```python
#операторы сравнения

# > больше
# < меньше
# >= больше либо равно
# <= меньше либо равно
# == равно
# != неравно
# isnull()
```

# Фильтрация строк between и isin

```python
#выводим первые 5 строк с датафрейма df_o
#формат данных серии OrderDate уже был нами ранее заменен на datetime64
df_o.head()
```

```python
#выводим заказы, чей вес входит в диапазон
#параметр inclusive: отвечает за включение границ. То есть, устанавливать ли каждую границу как закрытую или открытую.
#Может принимать значения: 'both', 'neither', 'left', 'right'.
#по умолчанию 'both', что означает обе границы включены
#При значении аргумента inclusive="neither" граничные значения исключаются
#значения 'left' или 'right' показывают, какая граница будет ВКЛЮЧЕНА в диапазон, если нужно оставить только одну
#inclusive='left' оставит в диапазоне левую границу, но иключит правую
df_o[df_o['Freight'].between(10,20,inclusive='neither')]
```

```python
#заказы, оформленные сотрудниками, чьи id не в списке
df_o[~(df_o['EmployeeID'].isin([2,4,6]))]
```

```python
#заказы клиентов, чьи ID в списке
df_o[df_o['CustomerID'].isin(['CHOPS','TORTU','RICSU'])]
```

# Фильтрация строк с and и or


```python
#выводим первые 5 строк с датафрейма df_p
df_p.head()
```

```python
#выводим товары категории Beverages, которые стоят больше 20
df_p[(df_p['CategoryName']=='Beverages') & (df_p['UnitPrice']>20)]
```

```python
#выводим данные с датафрейма df_o
#формат данных серии OrderDate уже был нами ранее заменен на datetime64
df_o.head()
```

```python
#заказы, оформленные в период с 1996-07-08 по 1996-10-06 сотрудником с id = 4
df_o[(df_o['OrderDate'].between('1996-07-08','1996-10-06'))&(df_o['EmployeeID']==4)]
```

```python
#заказы, оформленные в период с 1996-07-08 по 1996-10-06, или заказы, оформленные сотрудником с id = 4
df_o[(df_o['OrderDate'].between('1996-07-08','1996-10-06'))|(df_o['EmployeeID']==4)]
```

# Фильтрация по строковым данным

```python
#выводим первые 30 строк датафрейма
df_c.head(30)
```

```python
#находим должности, в которых присуствует слово Sales. Неважно где.
df_c[df_c['ContactTitle'].str.contains('Sales')]
```

```python
#Скорее всего Sales будет иметь разный регистр взависимости от положения в тексте (в реальной жизни).
#Поэтому предварительно можно перевести весть текст в нижний регистр.
df_c[df_c['ContactTitle'].str.lower().str.contains('sales')]
```

```python
#ищем должности, чьи названия начинаются с Sales
df_c[df_c['ContactTitle'].str.startswith('Sales')]
```

```python
#ищем должности, чьи названия заканчиваются на Manager
df_c[df_c['ContactTitle'].str.endswith('Manager')]
```

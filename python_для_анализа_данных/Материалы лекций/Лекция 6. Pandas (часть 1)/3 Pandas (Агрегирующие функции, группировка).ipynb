# Группиовка строк по одному столбцу

```python
#выводим 5 первых строк с датафрейма df_p
df_p.head()
```

```python
#считаем количество продуктов по каждой категории
# в groupby передаем название столбца по которому группируем
#получилась серия, где названия категорий - это индексы серии
df_p.groupby('CategoryName').ProductID.count()
```

```python
#переводим серию в датафрейм, а названия категорий из индексов в значения с помощью метода .reset_index()
df_p.groupby('CategoryName').ProductID.count().reset_index()
```

```python
#минимальная стоймость товара в каждой категории
df_p.groupby('CategoryName').UnitPrice.min().reset_index()
```

```python
#минимальная и максимальная стоймость товара в каждой категории
df_p.groupby('CategoryName').UnitPrice.agg(['min','max']).reset_index()
```

```python
#минимальная стоймость товара и количество товаров в каждой категории
df_p.groupby('CategoryName').agg({'UnitPrice':'min','ProductID':'count'}).reset_index()
```

```python
#выводим данные с датафрейма df_o
#у части заказов не указана дата доставки ShippedDate
df_o
```

```python
#кол-во заказов в разрезе дней доставки
# дополнительный параметр dropna= False позволяет посчитать количество заказов, у которых дата доставки пустая
df_o.groupby('ShippedDate',dropna= False).OrderID.count().reset_index()
```

```python
#выводим данные с датафрейма df_od
df_od
```

```python
#cчитаем фактическую скидку по каждой строке
df_od['Discount_fact'] = df_od['UnitPrice']*df_od['Quantity']*df_od['Discount']
```

```python
#считаем выручку очищенную от скидки по каждой строке
df_od['Revenue'] = df_od['UnitPrice']*df_od['Quantity']-df_od['Discount_fact']
```

```python
#появились 2 новых столбца
df_od
```

```python
#считаем фактическую скидку и чистую выручку каждому заказу. Записываем результат в новый датафрейм df_od_2
df_od_2 = df_od.groupby('OrderID').agg({'Discount_fact':'sum','Revenue':'sum'}).reset_index()
```

```python
#теперь выручка и скидка в разрезе каждого заказа, а кол-во заказов такое же как в df_o
df_od_2
```

```python
#переименовываем столбцы c помощью метода rename
#в методе rename в качестве параметра указываем, что собираемся переименовывать колонки (columns)
#и после символа '=' передаем словарь, в котором ключ - это старое название, значение - новое название
#пар ключ-значение может быть несколько
df_od_2 = df_od_2.rename(columns={'Discount_fact':'Discount','Revenue':'Clean_revenue'})
```

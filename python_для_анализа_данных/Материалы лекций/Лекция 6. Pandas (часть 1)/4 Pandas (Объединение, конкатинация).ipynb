Чуть-чуть подробнее про merge() и варианты её использования

Функция merge() в Pandas используется для объединения двух или более DataFrame на основе общих столбцов или индексов. Это аналогично операции соединения в базах данных.

Параметр функции how — это тип слияния. Возможные значения: 'left', 'right', 'outer', 'inner', 'cross'.

По умолчанию используется 'inner', которое **объединяет только общие строки** между двумя DataFrame.

'left' — использует все столбцы из левого DataFrame независимо от того есть ли соотношение по общему ключу в правой таблице.

'right' — использует все столбцы из правого DataFrame независимо от того есть ли соотношение по общему ключу в левой таблице.

'outer' — использует объединение всех столбцов из обоих DataFrame.

'inner' — вернёт из левой и правой таблиц только те строки, по которым нашлись соответствия.

'cross' — создает декартово произведение обоих DataFrame (все строки одного датафрейма пересекаются со всеми строками второго датафрейма)

Часто бывает, что ключи в таблицах называется по-разному. Например в датафрейме df_c ID клиента - это CustID, а в датафрейме df_o ID клиента - это CustomerID. В таком случае в методе .merge() нужно прописать дополнительные параметры left_on и right_on, где:

left_on - ключ левой таблицы (к которой присоединяем)

right_on - ключ правой таблицы (которую присоединяем)

```python
df_c.merge(df_o, left_on ='CustID', right_on = 'CustomerID')
```

Еще один **важнейший** нюанс:

В таблицах Products и Order_details есть общий ключ ProductID. Но также в этих таблицах есть еще один столбец с одинаковым именем UnitPrice. Причем UnitPrice в df_od - это стоимость продукта в момент оформления заказа, а UnitPrice в df_p - это стоимость продукта на «сегодняшний день». Стоимости продуктов в df_od и в df_p могут различаться.

Так как у нас 2 столбца, которые есть и в той и в другой таблице, этого кода будет недостаточно:

df_od.merge(df_p)

Потому что .merge()будет искать соответствия и по ProductID и по UnitPrice, что в свою очередь приведет к потере данных. Поэтому в методе .merge() необходимо явно прописать ключи.

```python
df_od.merge(df_p, left_on = 'ProductID', right_on = 'ProductID')
```

Теперь названия столбцов UnitPrice поменялись на UnitPrice_x и UnitPrice_y.

```python
# Переименовываем столбцы перед объединением
df_p_renamed = df_p.rename(columns={'UnitPrice': 'CurrentUnitPrice'})
df_od_renamed = df_od.rename(columns={'UnitPrice': 'OrderUnitPrice'})

# Объединяем по ProductID
merged_df = pd.merge(df_p_renamed, df_od_renamed, on='ProductID', how='inner')  # или другой тип соединения
merged_df
```

```python
# Переименовываем в процессе объединения
merged_df = pd.merge(
    df_p.rename(columns={'UnitPrice': 'CurrentPrice'}),
    df_od.rename(columns={'UnitPrice': 'OrderPrice'}),
    on='ProductID'
)
merged_df
```

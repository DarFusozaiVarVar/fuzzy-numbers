# Сводные таблицы

```python
#Обнуляем изменения, сразу переводим дату в нужный формат
df_o = pd.read_csv('/content/drive/MyDrive/Orders.csv', sep = ';', parse_dates=['OrderDate', 'RequiredDate', 'ShippedDate'])
```

```python
#вычленяем год, месяц, день месяца, квартал и неделю с даты заказа
df_o['Year'] = df_o['OrderDate'].dt.year
df_o['Month'] = df_o['OrderDate'].dt.month
df_o['Day'] = df_o['OrderDate'].dt.day
df_o['quarter'] = df_o['OrderDate'].dt.quarter
df_o['Week_ISO'] = df_o['OrderDate'].dt.isocalendar().week

#выводим 5 строк с датафрейма df_o
df_o.head()
```

```python
#сначала сгруппируем кол-во заказов по месяцу и году и сохраним отдельно этот датафрейм для дальнейшей работы
df_for_pivoting = df_o.groupby(['Year','Month']).OrderID.count().reset_index()

#проверяем что получилось
df_for_pivoting
```

```python
#теперь строим сводную таблицу
df_for_pivoting.pivot_table(index = 'Year', columns = 'Month', values = 'OrderID').reset_index()
#index - столбец, который расположится по вертикали
#columns - столбец, который расположится по горизонтали как название столбцов
#values - значения сводной таблицы
```

```python
#заменим пустые значения на 0 с помощью аргумента fill_value
df_for_pivoting.pivot_table(index = 'Year', columns = 'Month', values = 'OrderID', fill_value= 0).reset_index()
```

```python
#на самом деле, можно было не делать промежуточный датафрейм для сводной таблицы
#вернёмся к df_o, выводим 5 строк с датафрейма df_o
df_o.head()
```

```python
#можно было сделать группировку и агрегацию непосредственно в методе .pivot_table().
#для этого добавляем аргумент aggfunc, в который передаем название агрегирующей функции
df_o.pivot_table(index = 'Year', columns = 'Month', values = 'OrderID', fill_value= 0, aggfunc='count').reset_index()
```

#Категоризация непрерывных переменных

Признаки делятся на дискретные и непрерывные. Дискретные признаки могут принимать только определенные значения. Они могут быть:

-числовыми, например: количество заказов или ID продукта;

-в виде текста, например: название категории или название должности.

Непрерывные признаки не ограничиваются определенными отдельными значениями, а могут принимать любое значение в непрерывном диапазоне. Между любыми двумя непрерывными значениями может быть бесконечное число других. Непрерывные признаки всегда числовые. Примеры непрерывных признаков: стоимость товара, выручка, вес заказа.

Чуть выше мы разобрали один способ категоризировать дискретные признаки. мы исходя из обращения создали признак пол. Для этого мы создали вспомогательный датафрейм с столбцами: пол и обращение.

Теперь рассмотрим способ категоризировать числовые признаки, в том числе и непрерывные.

```python
#выводим 5 строк с df_p
df_p.head()
```

```python
#с помощью метода .cut() разбиваем стоймость на 4 равных бина (группы, отрезка). Добавляем новый столбец
df_p['bins']=pd.cut(df_p['UnitPrice'],4)

df_p
```

```python
#группируем по бинам
df_p.groupby('bins').ProductID.count()
```

```python
#видим, что из 77 товаров 73 приходятся на нижнюю группу цены
```

```python
#мы также можем сами формировать границы бинов
df_p['bins']=pd.cut(df_p['UnitPrice'], bins= [0,10,20,30,40])

#группируем по бинам
df_p.groupby('bins').ProductID.count()
```

```python
#стоит обратить внимание, что сюда попали не все товары

#в этом можно убедиться. Сортируем стоймость по убыванию и видим, что когда мы сами формируем диапазоны бинов, бины могут сформироваться не для всех цен
df_p.sort_values('UnitPrice',ascending = False)
```

```python
#добавляем названия бинов
df_p['bins']=pd.cut(df_p['UnitPrice'], bins= [0,10,20,30,40],labels=['от 1 до 10','от 10 до 20','от 20 до 30','от 30 до 40'])

df_p.groupby('bins').ProductID.count()
```

```python
#с помощью метода .qcut() разбиваем стоймость на 4 бина. При этом кол-во товаров в бинах стримится к одинаковому
df_p['bins']=pd.qcut(df_p['UnitPrice'], 4)

#группируем по бинам
df_p.groupby('bins').ProductID.count()
```

#Замена значений в датафреймах

```python
#обнулим предыдущие манипуляции с датафреймом с заказами
df_o = pd.read_csv('/content/drive/MyDrive/Orders.csv', sep = ';')

#выводим данные с df_o
df_o
```

```python
#смотрим есть ли пропуски
df_o.info()
```

```python
#видим, что количество значений ShippedDate меньше остальных. Но такой способ не самый информативный
```

```python
#метод .isnull() возращает True, если есть пропуск
df_o.isnull()
```

```python
#считаем кол-во пропусков
df_o.isnull().sum()
```

```python
#заменяем все пропуски столбца на дату с помощью метода .fillna()
df_o['ShippedDate'] = df_o['ShippedDate'].fillna('2025-11-24 11:55:00.000')

#проверяем
df_o
```

```python
#заменяем в столбце ShippedDate значение 1996-07-16 00:00:00.000 на notshipped с помошью метода .replace()
df_o['ShippedDate']= df_o['ShippedDate'].replace('1996-07-16 00:00:00.000','notshipped')
df_o
```

```python
# Метод where в pandas используется для выборочного сохранения или замены значений в DataFrame или Series на основе заданного условия.
# Он работает по принципу "сохранять, где условие истинно, заменять, где ложно"
```

```python
#заменяем значения столбца ShippedDate для заказов 11076,11077 на Notshipped
df_o['ShippedDate'] =df_o['ShippedDate'].where(~(df_o['OrderID'].isin([11076,11077])), 'Notshipped')

#проверяем
df_o
```

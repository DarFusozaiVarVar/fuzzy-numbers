# Метод apply()

```python
#считываем данные с файлов
df_p = pd.read_csv('/content/drive/MyDrive/Products.csv', sep = ';')
df_c = pd.read_csv('/content/drive/MyDrive/Customers.csv', sep = ';')
df_o = pd.read_csv('/content/drive/MyDrive/Orders.csv', sep = ';')
df_od = pd.read_csv('/content/drive/MyDrive/Order_details.csv', sep = ';')
df_e = pd.read_excel('/content/drive/MyDrive/Employees.xlsx')
```

```python
df_c.head()
```

Метод apply() в библиотеке Pandas позволяет применять пользовательскую функцию к каждому элементу или группе элементов в DataFrame или Series.

```python
df_c['Phone'].apply(lambda x: x.replace('-','').replace(' ','')) #удаляем из данных в столбце знаки пробела и '-'
```

```python
df_c['Phone'].str.replace('-','') #можно удалить знаки используя строчные методы
```

В качестве функции в методе apply() можно использовать не только лямбда-функции, но и любые другие.

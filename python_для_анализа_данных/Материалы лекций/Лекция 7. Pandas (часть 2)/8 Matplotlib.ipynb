Интерфейс pyplot позволяет пользователю сосредоточиться на выборе готовых решений и настройке базовых параметров рисунка. Это его главное достоинство, поэтому изучение matplotlib лучше всего начинать именно с интерфейса pyplot.

Рассмотрим простейший пример рисования графика с помощью
функций модуля pyplot. Создадим скрипт д ля рисования простого рисунка — графика
зависимости y = sin x. Для отображения набора пар значений x и y
используется функция plot() из модуля matplotlib.pyplot.

```python
import matplotlib.pyplot as plt
plt.plot(x, y, args) #на данный момент функция ничего не выводит, т.к. не заданы значения x / y
```

где x — массив значений, отображаемых по оси абсцисс (массив координат), y — массив значений, откладываемых по оси ординат (массив значений функции), args — перечисленные через запятую необязательные дополнительные аргументы для настройки графика.

Если в функцию plot() передаётся только один аргумент-массив, то в качестве массива координат для рисования графика будут использованы индексы элементов этого массива. Для изображения функциональной зависимости y = f(x) необхо-
димо создать массивы координат x = [x0, x1, …, xN] и массивы значений функции y = [y0, y1, …, yN] в соответствующих точках. Далее приведён текст программы для рисования графика зависимости y = sin (x). Массив координат создаётся с помощью функции linspace(), а массив значений функции — с помощью универсальной функции sin() из библиотеки NumPy. В конце программы вызывается функция show() из модуля pyplot, предназначенная для отображения рисунка на экране компьютера.

```python
import matplotlib.pyplot as plt
import numpy as np # подключение библиотеки numpy под псевдонимом np

x = np.linspace(0.0, 10.0, 30) # массив координат - 30 точек, равнораспределенных # в диапазоне от 0 до 10
y = np.sin(x) # массив значений функции в заданных координатах

plt.plot(x, y) # рисование графика функции с помощью функции plot
plt.show() # отображение рисунка на экране
```

Давайте разберем ещё 1 простой пример - график линейной зависимости. Как уже отмечалось, если передаётся только один аргумент-массив, то в качестве массива координат для рисования графика будут использованы индексы элементов этого массива, индексирование начинается с 0.

```python
import matplotlib.pyplot as plt
plt.plot([1,2,3,4])
plt.show()
```

Вам может не понравиться, что оси не начинаются с одного значения (например, с нуля, как в традиционном построении графиков функций в математике). Тогда вы можете задать значения осей вручную с помощью метода **axis**:


```python
import matplotlib.pyplot as plt
plt.plot([1,2,3,4])
plt.axis([0, 4, 0, 5]) #начало и конец оси Х, начало и конец оси Y
plt.show()
```

Для каждой пары аргументов x, y есть необязательный третий аргумент, который представляет собой строку формата, указывающую цвет и тип линии графика. Буквы и символы строки формата взяты из MATLAB, и вы конкатенируете строку цвета со строкой типа линии. По умолчанию используется строка формата 'b-', которая представляет собой сплошную синюю линию. Например, чтобы построить график с красными кругами, нужно ввести команду

```python
import matplotlib.pyplot as plt

plt.plot([1,2,3,4], [1,4,9,16], 'ro')
plt.axis([0, 6, 0, 20])
plt.show()
```

Пример ниже иллюстрирует построение нескольких линий с разными стилями формата в одной команде с использованием массивов:

```python
import numpy as np
import matplotlib.pyplot as plt

# evenly sampled time at 200ms intervals
t = np.arange(0., 5., 0.2)

# red dashes, blue squares and green triangles
plt.plot(t, t, 'r--') #красный пунктир
plt.plot(t, t**2, 'bs') #синий квадрат
plt.plot(t, t**3, 'g^') #зеленый треугольник
plt.show()
```

Линии имеют множество атрибутов, которые вы можете задать: ширина линии, стиль штрихов, сглаживание и т. д. (подробности можно посмотреть в приведенной выше документации). Существует несколько способов задать свойства линии, например, используя ключевые слова в аргументах:

```python
import matplotlib.pyplot as plt
plt.plot([1,2,3,4], linewidth=3.0)
plt.plot([4,3,2,1], linewidth=8.0, color='r') #обратите внимание, что этот график перекроет предыдущий, т.к. был вызван позже
plt.axis([0, 4, 0, 5])
plt.show()
```

### 1.3 Создание графиков с несколькими областями

Все объекты в Matplotlib имеют иерархическую (древоподобную) структуру. Основными элементами являются рисунок (верхняя иерархия), панели (средняя иерархия) и оси координат (нижняя иерархия).

На верхнем уровне иерархии находится окно рисунка, которое является контейнером для остальных элементов. Рисунок является объектом класса Figure.

На рисунке может располагаться несколько (одна или больше) панелей — независимых областей рисования. Панели я вляются объектами класса Axes.

Каждая из панелей содержит основные элементы рисунка: оси координат, метки на осях, текстовые метки, графики и т. д.

Алгоритм работы с рисунками можно записать в виде следующей последовательности действий:

1. Создание окна рисунка. Окно создаётся с помощью функции matplotlib.pyplot.figure() из модуля pyplot. Например, конструкция вида:

```python
fig = plt.figure(args)
```

создаёт окно рисунка и ассоциирует его с переменной fig. Данная переменная является объектом класса Figure. Настройки окна в дальнейшем осуществляются посредством обращения к полям и методам этого объекта. Функция figure() может принимать необязательные аргументы для настройки окна рисунка. Например,
конструкция

```python
fig = plt.figure(figsize=(width, height))
```

создаёт рисунок шириной width и высотой height (в дюймах). Будем в дальнейшем использовать имя fig в к ачестве и мени по умолчанию для создаваемого окна.

2. Добавление в окно панелей (подграфиков) для рисования графиков. Одним из способов добавления панели на рисунок является использование метода add_subplot(nrows, ncols, index) объекта «рисунок». Три целочисленных аргумента указывают, что панель займёт на рисунке положение под номером index в таблице из nrows строк и ncols столбцов. Нумерация начинается с 1 в левом верхнем углу таблицы и увеличивается слева направо. Например, конструкция вида:

```python
ax = fig.add_subplot(1,1,1)
```

добавляет в ранее созданное окно fig одну панель и ассоциирует её с переменной ax.

Давайте рассмотрим несколько практических примеров, чтобы стало понятнее, как появляются подграфики на фигуре.

```python
#Пустой рисунок matplotlib с тремя подграфиками
import matplotlib.pyplot as plt
fig = plt.figure()

ax1 = fig.add_subplot(2, 2, 1) #рисунок будет расчерчен сеткой 2×2, и мы выбираем первый из четырех подграфиков (нумерация начинается с 1)
ax2 = fig.add_subplot(2, 2, 2) #создать следующие два подграфика
ax3 = fig.add_subplot(2, 2, 3)
```

Эти объекты графика обладают различными методами, которые создают графики различных типов. Например, можно было бы построить линейный график методом plot

```python
import matplotlib.pyplot as plt
import numpy as np

fig = plt.figure()

ax1 = fig.add_subplot(2, 2, 1)
ax2 = fig.add_subplot(2, 2, 2)
ax3 = fig.add_subplot(2, 2, 3)

ax3.plot(np.random.standard_normal(50).cumsum(), color="black", linestyle="dashed") #создаем случайный график в подграфике 3 с заданными дополнительными параметрами, которые обсудим ниже
```

Возможно, вы обратили внимание на сообщение <matplotlib.lines.Line2D at ...>, появившееся после выполнения этого метода. Matplotlib возвращает объекты, ссылающиеся на только что добавленную компоненту графика. Чаще всего результат можно спокойно игнорировать, но можно поставить ; в конце строки с вызовом графика, чтобы "подавить" вывод.

Дополнительные параметры говорят matplotlib, что график должен быть на-
рисован черной пунктирной линией.

Давайте добавим графики в другие окна панелей.

```python
import matplotlib.pyplot as plt
import numpy as np

fig = plt.figure()

ax1 = fig.add_subplot(2, 2, 1)
ax2 = fig.add_subplot(2, 2, 2)
ax3 = fig.add_subplot(2, 2, 3)

ax3.plot(np.random.standard_normal(50).cumsum(), color="black", linestyle="dashed")
ax1.hist(np.random.standard_normal(100), bins=20, color="black", alpha=0.3) #параметр стиля alpha=0.3 задает степень прозрачности наложенного графика.
ax2.scatter(np.arange(30), np.arange(30) + 3 * np.random.standard_normal(30));
```

Ещё пример создания нескольких графиков на одном рисунке:

```python
import numpy as np
import matplotlib.pyplot as plt

def f(t):
    return np.exp(-t) * np.cos(2*np.pi*t)

t1 = np.arange(0.0, 5.0, 0.1)
t2 = np.arange(0.0, 5.0, 0.02)

plt.figure(1)
plt.subplot(211)
plt.plot(t1, f(t1), 'bo', t2, f(t2), 'k')

plt.subplot(212)
plt.plot(t2, np.cos(2*np.pi*t2), 'r--')
plt.show()
```

### 1.4 Работа с текстом на графиках

Команда text() может использоваться для добавления текста в произвольное место, а команды xlabel(), ylabel() и title() используются для добавления текста в указанные места (оси х, y и название соответственно).



```python
import numpy as np
import matplotlib.pyplot as plt

#код для создания разброса с широкой выборкой
np.random.seed(19680801)
mu, sigma = 100, 15
x = mu + sigma * np.random.randn(10000)
n, bins, patches = plt.hist(x, 50, facecolor='g', alpha=0.75)


plt.xlabel('Уровень IQ')
plt.ylabel('Вероятность')
plt.title('Гистограмма IQ')
plt.text(50, 400, r'$\mu=100,\ \sigma=15$')

plt.grid(True)
plt.show()
```

Как и в случае с линиями выше, вы можете настроить свойства, передавая аргументы ключевых слов в функции text:

```python
import matplotlib.pyplot as plt

plt.plot([4,3,2,1], linewidth=8.0, color='r') #обратите внимание, что этот график перекроет предыдущий, т.к. был вызван позже
plt.axis([0, 4, 0, 5])

plt. title('Клёвая красная линия', fontsize=14, color='red')
plt.show()
```

**Использование математических выражений в тексте**

На рисунки matpotlib можно добавлять математические надписи. Набор математических выражений осуществляется по правилам системы ТеХ (TEX — созданная американским математиком и программистом Дональдом Кнутом система для верстки текстов с математическими формулами.). Перед строкой, содержащей математическое выражение, помещается буква r, а внутри кавычек используются знаки доллара. Вы могли заметить математические символы на графике с гисторграммой ранее.

**Аннотации**

Приведенные выше варианты использования базовой команды text() размещают текст в произвольной позиции на осях. Чаще всего текст используется для аннотирования некоторых элементов графика, и метод annotate() предоставляет вспомогательную функциональность для упрощения аннотирования. В аннотации необходимо учитывать два момента: местоположение аннотируемого объекта, представленное аргументом xy, и местоположение текста xytext. Оба аргумента представляют собой кортежи (x,y).

```python
import numpy as np
import matplotlib.pyplot as plt



t = np.arange(0.0, 5.0, 0.01)
s = np.cos(2*np.pi*t)
line, = plt.plot(t, s, lw=2)

plt.annotate('local max', xy=(2, 1), xytext=(3, 1.5), arrowprops=dict(facecolor='black', shrink=0.05))

plt.ylim(-2,2)
plt.show()
```

В этом базовом примере координаты xy (наконечник стрелки) и xytext (местоположение текста) указаны в координатах данных. Существует множество других систем координат, которые можно выбрать. Подробнее с ними можно ознакомиться в приведенной в начале документации.

**Добавление пояснительных надписей**

Пояснительная надпись – еще один важный элемент оформления графика. До-
бавить ее можно двумя способами. Проще всего передать аргумент label при
добавлении каждого нового графика:

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, label='sin(x)')
plt.xlabel('x')
plt.ylabel('y')
plt.title('График синуса')
plt.legend()

plt.show()
```

### 1.5 Логарифмические и другие нелинейные оси

matplotlib.pyplot поддерживает не только линейные масштабы осей, но и логарифмические и логарифмические масштабы. Это обычно используется, если данные охватывают много порядков величины. Изменить масштаб оси очень просто:

**plt.yscale('log')**



```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(1, 10, 100)
y = np.exp(x)

plt.plot(x, y)
plt.yscale('log')
plt.title('График функции y = e^x в логарифмическом масштабе')
plt.xlabel('x')
plt.ylabel('y')

plt.show()
```

### 1.6 Диаграммы рассеивания и трехмерные графики

**Диаграммы рассеивания**

Функция plt.scatter() в библиотеке Matplotlib используется для построения диаграммы рассеяния. Эта диаграмма позволяет оценить взаимосвязь между двумя переменными, значения которых откладываются по разным осям.

Вот основные параметры функции `plt.scatter()`:

- `x`: Список или массив значений по оси X.
- `y`: Список или массив значений по оси Y.
- `c`: Цвет или последовательность цветов для окраски точек. Может быть строкой, определяющей цвет, или последовательностью значений в формате RGB (например, 'r', 'g', 'b', 'c', 'm', 'y', 'k', 'w' и т.д.).
- `marker`: Форма маркера, используемая для обозначения точек на графике (например, 'o', '^', 's', '*', 'x', 'D' и т.д.).
- `label`: Подпись для легенды.
- Другие параметры для настройки внешнего вида точек, такие как размер, прозрачность и т.д.

Код ниже сгенерирует 1000 случайных точек на плоскости и присвоит каждой точке цвет в зависимости от ее расстояния от начала координат. Полученный график будет показывать распределение точек с использованием цветовой карты (цветового ряда). Цветовая карта «viridis» — одна из четырёх новых однородных цветовых карт, представленных в matplotlib версии 2.0. Она обеспечивает высокую точность и доступность визуализации данных, минимизируя визуальные искажения.

```python
import numpy as np
import matplotlib.pyplot as plt

N = 1000  # Количество точек
x = np.random.rand(N)  # Генерируем случайные значения для координат x
y = np.random.rand(N)  # Генерируем случайные значения для координат y
colors = np.sqrt(x**2 + y**2)  # Вычисляем значения для цветового ряда

plt.scatter(x, y, c=colors, cmap='viridis', alpha=0.7)  # Строим точки, цвета которых зависят от расстояния от начала координат
plt.colorbar()  # Добавляем цветовую шкалу

plt.xlabel('X')
plt.ylabel('Y')
plt.title('Распределение точек по плоскости с цветовым рядом')

plt.show()
```

**Создание трехмерных графиков**

Для создания трехмерных графиков в Python с использованием модуля matplotlib, следуйте следующим шагам:

1. Импорт необходимых библиотек: импортируйте модуль matplotlib и подмодуль mpl_toolkits.mplot3d для работы с трехмерными графиками

2. Создание фигуры (области для рисунка) и осей в трехмерном пространстве.

3. Построение трехмерного графика: методы для построения трехмерных графиков, например, scatter, plot, bar, и т.д. В данном случае для поверхности использован plot_surface

4. Настройка графика:оси, метки, заголовки и другие компоненты графика, как в обычных двумерных графиках


```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Создаем данные
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
x, y = np.meshgrid(x, y)
z = np.sin(np.sqrt(x**2 + y**2))

# Создаем рисунок
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Построение поверхности
ax.plot_surface(x, y, z, cmap='viridis')

# Добавляем метки
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')

plt.title('Трехмерный график функции sin(sqrt(x^2 + y^2))')

plt.show()
```

### 1.7 Сохранение графиков в файле

Чтобы сохранить график из matplotlib в файл, можно использовать метод savefig() объекта Figure.

Вот пример кода, который сохранит график в формате PNG:

```python
import numpy as np
import matplotlib.pyplot as plt

# Генерация данных для графика
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Построение графика
plt.plot(x, y)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Простой график синуса')

# Сохранение графика в файл формата PNG
plt.savefig('simple_plot.png')

plt.show()
```

После выполнения этого кода, в вашей рабочей директории будет создан файл simple_plot.png, содержащий сохраненный простой двумерный график.

Формат выводится из расширения имени файла. Если бы мы задали файл
с расширением .pdf, то рисунок был бы сохранен в формате PDF. При публикации графики часто используют параметр dpi – разрешение в точках на дюйм.
Чтобы получить тот же самый график в формате PNG с разрешением 400 DPI,
нужно было бы написать:

*plt.savefig('simple_plot.png', dpi=400)*

### 1.8 Визуализация данных из текстовых файлов

Для рисования графиков по данным из текстового файла можно пользоваться стандартным методом чтения данных из файла:

```python
import matplotlib.pyplot as plt

# Чтение данных из текстового файла
with open('data.txt', 'r') as file:
    data = file.readlines()

# Разделение данных на оси X и Y
x = []
y = []
for line in data:
    values = line.split(',')
    x.append(float(values[0]))
    y.append(float(values[1]))

# Построение графика
plt.plot(x, y, marker='o', color='b', linestyle='-')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('График данных из файла')
plt.grid(True)
plt.show()
```

Кроме того, для рисования графиков по данным из текстового файла удобно
воспользоваться командой loadtxt() из библиотеки NumPy:

```python
import numpy as np
import matplotlib.pyplot as plt

# Чтение данных из текстового файла с помощью NumPy
data = np.loadtxt('data.txt', delimiter=',')

# Разделение данных на оси X и Y
x = data[:, 0]
y = data[:, 1]

# Построение графика
plt.plot(x, y, marker='o', color='b', linestyle='-')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('График данных из файла с использованием NumPy')

plt.show()
```

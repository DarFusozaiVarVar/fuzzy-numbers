**1 Общая стоимость**

Вам доступен текстовый файл prices.txt с информацией о заказе из интернет магазина. В нем каждая строка с помощью символа табуляции (\t) разделена на три колонки:

* наименование товара;
* количество товара (целое число);
* цена (в рублях) товара за 1 шт (целое число).


Напишите программу, выводящую на экран общую стоимость заказа.

```python
#результат для проверки - 1286800

def total_price(file):
    total = 0
    opened_file = open(file, 'r', encoding = 'utf-8')
    for line in opened_file.readlines():
        line = line.strip().split()
        item, count, price = line
        total += int(count) * int(price)
    return total
    
print(total_price('prices.txt'))
```

**4 Сумма чисел в файле**

Вам доступен текстовый файл nums.txt. В файле могут быть записаны целые неотрицательные числа и все что угодно. Числом назовем последовательность одной и более цифр, идущих подряд (число всегда неотрицательно).

Напишите программу, которая вычисляет сумму всех чисел, записанных в файле.



```python
# Результат для проверки - 124410
file = open('nums.txt', 'r')
lines = file.readlines()
file.close()
sm = 0
for l in lines:
    s = ''
    for i in l:
        if i.isdigit():
            s += i
        else:
            if s != '':
                sm += int(s)
            s = ''
print(sm)
```

**10 Популярные домены**

Вам доступен файл data.csv, который содержит неповторяющиеся данные о пользователях некоторого ресурса. В первом столбце записано имя пользователя, во втором — фамилия, в третьем — адрес электронной почты:

```python
""" first_name,surname,email
John,Wilson,johnwilson@outlook.com
Mary,Wilson,marywilson@list.ru
... """

import csv

with open('data.csv', encoding='utf-8') as file:
    rows = csv.reader(file, delimiter=',', quotechar='"')
    domen = {}
    for r in rows:
        try:
            dom = r[2].split('@')[1]
            if dom in domen:
                domen[dom] += 1
            else:
                domen[dom] = 1
        except:
            pass

    domen = [c for c in domen.items()]
    domen = sorted(domen, key=lambda x: (x[1], x[0]))
    print('domain, count')
    for line in domen:
        print(line[0] + ', ' + str(line[1]))
```

Напишите программу, которая выводит следующие данные:

```python
domain,count
rambler.ru,24
iCloud.com,29
...
```

где в первом столбце записано название почтового домена, а во втором — количество пользователей, использующих данный домен. Домены должны быть расположены в порядке возрастания количества их использований, при совпадении количества использований — в лексикографическом порядке.

**11 Сортировка по столбцу**

Вам доступен файл deniro.csv, каждый столбец которого содержит либо только числа, либо строковые значения:

```python
Machete,2010,72
Marvin's Room,1996,80
Raging Bull,1980,97
...
```

```python
import csv

with open('deniro.csv', encoding='utf-8') as file:
    rows = csv.reader(file, delimiter=',', quotechar=' ')
    n = int(input())
    r = list(rows)
    print(r)
    
    for i in r:
        for e in range(0, len(i)):
            if i[e].isdigit():
                i[e] = int(i[e])
    
    r = sorted(r, key = lambda x: (isinstance(x[n-1], str), x[n-1]))
    print(r)

    
```

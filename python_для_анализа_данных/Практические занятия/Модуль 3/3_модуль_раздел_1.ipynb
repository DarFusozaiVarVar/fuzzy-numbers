**Задача 6.**

Предположим, у вас есть набор данных о продажах товаров, представленный в виде двумерного массива sales_data_array. Каждая строка представляет собой информацию о продажах для конкретного товара, а столбцы содержат следующую информацию: номер товара, количество проданных единиц, и доход от продаж.

Вам необходимо написать функцию analyze_sales_data, принимающую массив sales_data и возвращающую 2 булевых значения в ответ на следующие вопросы:

* Есть ли хотя бы одна запись с нулевыми продажами (количество проданных единиц равно 0)?
* Все ли товары имеют положительный доход от продаж?

```python
import numpy as np
def analyze_sales_data(sales_data):
    ans = (np.isin(0, sales_data[:, 1:2]), np.isin(True, np.where((sales_data[:, 1:2]) < 0, True, False), invert=True))
    return ans

sales_data_array = np.array([[1, 10, 100],
                             [2, 5, 50],
                             [3, 0, 0],
                             [4, 8, 80]])

ans = analyze_sales_data(sales_data_array)
print(f"Есть ли товары с нулевыми продажами: {ans[0]}")  # первое булево значение согласно условию
print(f"Все ли товары имеют положительный доход: {ans[1]}")  # второе булево значение согласно условию
```

**Задача 8.**

У вас есть два массива: array1 = np.array([10, -15, 8, -20]) и array2 = np.array([3, 4, 2, 7]). Напишите программу для копирования знаков значений из второго массива в значения первого массива.

```python
import numpy as np

array1 = np.array([10, -15, 8, -20])
array2 = np.array([-3, 4, -2, 7])

result = np.copysign(array1, array2)
print(result)
# нужный вывод:
# [-10.  15.  -8.  20.]
```

**Задача 12.**

Решите систему уравнений средствами NumPy:

3x + 2y - z = 10

x - y + 2z = 5

2x + 3y + z = 8

```python
Решение системы:
  x = 3.55
  y = 0.05
  z = 0.75
```

```python
import numpy as np

A = np.array([[3, 2, -1],
              [1, -1, 2],
              [2, 3, 1]])
b = np.array([10, 5, 8])

sol = np.linalg.solve(A, b)
x, y, z = sol
print(f"x = {x:.2f}, y = {y:.2f}, z = {z:.2f}")
```
